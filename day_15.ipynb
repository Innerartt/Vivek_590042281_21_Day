#Built in Modules and Exceptions

```python
import math as m
print(m.pi)
print(m.factorial(5))
```

```python
import random
random.seed(7)
print(random.randint(1,10))
```

```python
from datetime import date
today=date.today()
print(today.year,today.month,today.day)
```

```python
try:
    result=10/2
    print(result)
except ZeroDivisionError:
    print("cannot divide by zero")
```

```python
try:
    number=int("42")
    print(number)
except ValueError:
    print("Invalid Number")
```

```python
try:
    print([1,2][5])
except IndexError:
    print("Index is outside the list")
```

```python
try:
    value=int("abc")
except ValueError:
    value=0
finally:
    print("Finished",value)
```

```python
def safe_divide(a,b):
    try:
        return a/b
    except ZeroDivisionError:
        return None
print(safe_divide(10,0))
```

```python
numbers=["10","20","bad","30"]
valid=[]
for item in numbers:
    try:
        valid.append(int(item))
    except ValueError:
        pass
print(valid)
```

```python
def positive_Number(value):
    if value<0:
        raise ValueError("Value must be positive")
    return value
print(positive_Number(5))
```
