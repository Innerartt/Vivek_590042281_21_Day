#Nested Loops

```python
for row in range(3):
    for coloumn in range(3):
        print(row,coloumn)
```

```python
for number in range(1,4):
    for multiplier in range(1,4):
        print(number*multiplier , end=" ")
    print()
```

```python
for row in range(1,5):
    print("*" *row)
```

```python
matrix=[[1,2],[3,4]]
for row in matrix:
    for value in row:
        print(value)
```

```python
matrix=[[1,2],[3,4]]
total=0
for row in matrix:
    for value in row:
        total+=value
print(total)
```

```python
for first in range(1,4):
    for second in range(first,4):
        print(first,second)
```

```python
words=["hi","bye"]
for word in words:
    for letter in word:
        print(letter)
```

```python
for row in range(3):
    line=" "
    for coloumn in range(5):
        line+="#"
    print(line)
```

```python
coordinates=[(x,y) for x in range(10) for y in range(10)]
print(coordinates)
```

```python
for number in range(2,10):
    factors=[]
    for candidate in range(1,number):
        if number%candidate==0:
            factors.append(candidate)
    print(number,factors)
```
