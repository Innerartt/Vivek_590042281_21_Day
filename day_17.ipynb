#Small Number Codes

```python
number=7
if number%2==0:
    print("even")
else:
    print("odd")
```

```python
number=5
fact=1
for i in range(1,number+1):
    fact=fact*i
print("the factorial of the number is:",fact)
```

```python
first,second=0,1
for i in range(8):
    print(first,end=" ")
    first,second=second,first+second
```

```python
number=12345
print(sum(int(digit) for digit in str(number)))
```

```python
number=12345
print(int(str(number)[::-1]))
```

```python
number=29
divisor=[value for value in range(1,number+1) if number%value==0]
print(divisor)
```

```python
number=[12,5,8,20,3]
print(sum(number),sum(number)/len(number))
```

```python
number=153
digits=[int(digit) for digit in str(number)]
print(sum(digit**3 for digit in digits)==number)
```

```python
def gcd(a,b):
    while b:
        a,b=b,a%b
    return a
print(gcd(48,18))
```

```python
numbers=[4,7,2,9,1]
print(sorted(numbers,reverse=True)[:2])
```
