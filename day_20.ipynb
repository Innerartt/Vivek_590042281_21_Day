```python
students=[{'name':'Asha','score':[80,90]},{'name':'Ravi','score':[70,75]}]
for student in students:
    print(student['name'],sum(student['score'])/len(student['score']))
```

```python
def grade(score):
    if score>=90:
        return 'A'
    if score>=75:
        return 'B'
    return 'C'
print([grade(score) for score in [92,81,60]])
```

```python
cart={'rice':2,'tea':1,'soap':3}
prices={'rice':60,'tea':120,'soap':35}
total=sum(cart[item]*prices[item] for item in cart)
print(total)
```

```python
def invoices(items):
    subtotal=sum(item['price']*item['quantity'] for item in items)
    tax=subtotal*0.18
    return subtotal,tax,subtotal+tax

print(invoices([{'price':100,'quantity':2},{'price':50,'quantity':1}]))
 
```

```python
numbers=[12,7,18,3,20]
result={'even': [n for n in numbers if n%2 == 0], 'odd': [n for n in numbers if n%2 !=0]}
print(result)
```

```python
def word_summary(text):
    words=text.lower().split()
    return {'words':len(words),'unique':len(set(words)),'longest':max(words,key=len)}
print(word_summary('python makes learning simple'))
```

```python
account={'balance':1000}
transactions=[250,-100,-75,300]
for transaction in transactions:
    account['balance']+=transaction
print(account)
```

```python
def find_names(records,minimum_score):
    return [record['name'] for record in records if record['score']>=minimum_score]
records=[{'name':'Asha','score':88},{'name':'Ravi','score':72}]
print(find_names(records,80))
```

```python
inventory={'pen':10,'book':4}
orders=[('pen',3),('book',2)]
for item,quantity in orders:
    if inventory.get(item,0)>=quantity:
        inventory[item]-=quantity
print(inventory)
```
