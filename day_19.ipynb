#Menu Style Programs

```python
def add(a,b):
    return a+b
add(4,7)
```

```python
def calculate(a,b, operation):
    if operation=="+":
        return a+b
    elif operation=="-":
        return a-b
    elif operation=="*":
        return a*b
    else:
        return "Unknown Operation"
print(calculate(8,3,"*"))
```

```python
menu={1:"Add",2:"Subtract",3:"Exit"}
for number,label in menu.items():
    print(number,label)
```

```python
choice=2
if choice==1:
    print("Add Selected")
elif choice==2:
    print("Subtract Selected")
else:
    print("Exit Selected")
```

```python
def convert_currency(amount,rate=83):
    return amount*rate
print(convert_currency(10))
```

```python
def ticket_price(age):
    if age<5:
        return 0
    elif age<18:
        return 50
    return 100
print(ticket_price(10))
```

```python
def loan_status(income,score):
    if income>=500000 and score>=700:
        return 'Approved'
    return 'Review'
print(loan_status(600000,720))
```

```python
def make_report(name,scores):
    average=sum(scores)/len(scores)
    status='Pass' if average >=50 else "Needed practice"
    return f'{name}:{average:.1f}-{status}'
print(make_report("Asha",[80,75,90]))
```

```python
def apply_discount(price,customer_type):
    rates={'regular':0.05,'member':0.15}
    return price*(1-rates.get(customer_type,0))
print(apply_discount(1000,'member'))
```
