#Validation with Control Flow

```python
value=25
print(1<=value<=100)
```

```python
email="user@example.com"
valid="@" in email and "." in email
print(valid)
```

```python
password="python123"
if len(password)>=8:
    print("Strong Password")
else:
    print("Weak Password")
```

```python
numbers=[2,4,6,8]
print(all(number%2==0 for number in numbers ))
```

```python
values=[3,7,0,9]
print(any(value==0 for value in values ))
```

```python
def valid_age(age):
    return isinstance(age,int) and 0<= age<=120
print(valid_age(120),valid_age(-2))
```

```python
choices =["yes","no"]
answer="yes"
if answer.lower() in choices:
    print("Valid Choice")
```

```python
records=[{"name":"Asha","score":85},{"name":"Ravi","score":72}]
for record in records:
    if 0<=record["score"]<=100:
        print(record["name"],"valid")
```

```python
number=29
is_prime=number>1
for divisor in range(2,number):
    if number%divisor==0:
        is_prime=False
        break
print(is_prime)
```

```python
required=["name","email"]
profile={"name":"Mira","email":"mira@example.com"}
print(all(field in profile for field in required))
```
