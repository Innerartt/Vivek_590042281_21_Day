#Lambda Function and Comprehensions

```python
double=lambda number:number*2
print(double(6))
```

```python
add = lambda a,b:a+b
print(add(3,5))
```

```python
numbers=[1,2,3,4]
print(list(map(lambda x:x**2,numbers)))
```

```python
numbers=[1,2,3,4]
print(list(filter(lambda x:x%2,numbers)))
```

```python
names=["Ravi","Asha","Mira"]
print(sorted(names,key=lambda name:len(name)))
```

```python
squares=[number**2 for number in range(1,6)]
print(squares)
```

```python
even_number=[number for number in range(10) if number %2==0]
print(even_number)
```

```python
labels=["even" if number%2==0 else "odd" for number in range(1,10)]
print(labels)
```

```python
length={word:len(word) for word in ["cat","python","book"]}
print(length)
```

```python
pairs=[(x,y) for x in range(1,3) for y in range(1,3)]
print(pairs)
```
