#Basic Data Processing 


```python
sales=[120,100,110,140,150]
print(sum(sales))
```

```python
sales=[120,100,110,140,150]
print(max(sales))
print(min(sales))
```

```python
sales=[120,90,150,80]
sales_over_100=[]
for value in sales:
    if value >=100:
        sales_over_100.append(value)
print(sales_over_100)
```

```python
items=[{'name':'Pen','price':100},{'name':'Book','price':120}]
print(sum(item['price'] for item in items ))
```

```python
employees=[{"name":"Asha","score":91},{"name":"Ravi","score":78}]
print(sorted(employees,key=lambda employee : employee['score'],reverse=True))
```

```python
def summarize(values):
    return {'count':len(values),"total":sum(values),"average":sum(values)/len(values)}
print(summarize([10,20,30]))
```

```python
prices=[100,250,80]
discounted=[price*0.9 for price in prices]
print(discounted)
```

```python
words=['apple','banana','apple','orange']
counts={word: words.count(word) for word in set(words)}
print(counts)
       
```

```python
scores={'Asha':91,"Ravi":78,"Mira":88}
passed={name : score for name,score in scores.items() if score>=80}
print(passed)
```

```python
transactions=[100,-40,250,-39]
balance=0
for transaction in transactions:
    balance+=transaction
print(balance)
```
