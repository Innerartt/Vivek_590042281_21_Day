#Function and Scope of variable

```python
message="outside"
def show_message():
    print(message)
show_message()
```

```python
def make_message():
    message="inside"
    return message
print(make_message())
```

```python
def divide(first,second):
    return first//second,first%second
quotient,remainder=divide(17,5)
print(quotient,remainder)
```

```python
def min_max(values):
    return(min(values),max(values))
print(min_max([4,9,1]))
```

```python
def operations(a,b):
    return(a+b,a*b)
print(operations(3,4))
```

```python
tax_rate=0.18
def price_with_tax(price):
    return price*(1 + tax_rate)
print(price_with_tax(500))
```

```python
def counter():
    count=0
    count+=1
    return count
print(counter())
```

```python
def format_person(person):
    return f"{person['name']} is {person['age']} years old"
print(format_person({"name":"Mira","age":22}))
```

```python
def calculate_bill(items):
    subtotal=sum(item["price"] for item in items)
    return subtotal,subtotal*1.18
print(calculate_bill([{"price":100},{"price":50}]))
```

```python
def describe(value):
    if value is None:
        return "No Value"
    return f'Value:{value}'
print(describe(None))
print(describe(10))
```
