```python
def reverse_words(sentence):
    return " ".join(sentence.split()[::-1])
print(reverse_words("learn by doing"))
```

```python
def remove_spaces(text):
    return text.replace(" ","")
print(remove_spaces("data science"))
```

```python
def count_characters(text):
    counts={}
    for character in text.lower():
        if character.isalpha():
            counts[character]=counts.get(character,0)+1
    return counts
print(count_characters("banana"))
```

```python
words=["cat","elephant","dog"]
print(max(words,key=len))
```

```python
text="hello world"
print(''.join(word[0].upper() for word in text.split()))
```

```python
def mask_email(email):
    name,domain=email.split("@")
    return_name=name[0]+"*"*(len(name)-1)
    return f'{return_name}@{domain}'
print(mask_email("student@example.com"))
```
