```python
def average(values):
    return sum(values)/len(values)
print(average([10,20,30]))
```

```python
def is_pallindrome(text):
    cleaned=text.lower().replace(' ','')
    return cleaned==cleaned[::-1]
print(is_pallindrome("Never odd or even"))
```

```python
def fibbionaci(count):
    values=[]
    first,second=0,1
    for i in range(count):
        values.append(first)
        first,second=second,first+second
    return values
print(fibbionaci(10))
```

```python
def frequency_table(text):
    table={}
    for word in text.lower().split():
        table[word]=table.get(word,0)+1
    return table
print(frequency_table('learn code learn daily'))
```

```python
def classify_numbers(numbers):
    return{'positive' :[n for n in numbers if n>0],'negative':[n for n in numbers if n<0],'zero':numbers.count(0)}
print(classify_numbers([-2,0,4,-1,5]))
```

```python
def grade_report(records):
    report={}
    for name,score in records.items():
        report[name]='pass' if score>=50 else "fail"
    return report
print(grade_report({'Asha':82,"Ravi":45}))
```

```python
def shopping_total(cart,prices):
    return sum(quantity*prices[item] for item,quantity in cart.items())
print(shopping_total({'book':2,'pen':3},{'book':80,'pen':10}))
```

```python
numbers=list(range(1,21))
primes=[]
for number in numbers:
    if number>1 and all(number%divisor for divisor in range(2,number)):
        primes.append(number)
print(primes)
```

```python
def summarize(scores):
    return{'highest':max(scores),'lowest':min(scores),'average':round(sum(scores)/len(scores),2),'passed':sum(score >=50 for score in scores )}
print(summarize([45,72,88,91,36]))
```
