# #While Loops


```python
count=1
while count <=5:
    print(count)
    count+=1
```

```python
count=10
while count >0:
    print(count)
    count=count-2
```

```python
total=0
number =1
while number <=5:
    total+=number
    number+=1
print(total)
```

```python
string="python"
number=0
while number< len(string):
    print(string[number])
    number+=1
```

```python
items= ["Python","helloworld","bag"]
number=0
while number < len(items):
    items.pop()
print("the item list is :",items)
```

```python
attempts=0
while attempts<3:
    print("attempts",attempts+1)
    attempts+=1
```

```python
a=int(input("enter a number: "))
while True:
    if a>=5:
        break
    print(a)
    a+=1
print("the number inputed is: ",a)
```

```python
number=0
while number<5:
    if number==3:
        number+=1
        continue
    print(number)
    number+=1
```

```python
balance=100
day=0
while balance<150:
    balance+=10
    day+=1
print(day,balance)
```
